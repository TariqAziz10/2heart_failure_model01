```python
#how many country have Republic in their full name
count = 0
def extracting_name(txt):
  global count
  if 'republic' in txt.lower():
    count += 1
  return txt

df['country_long'] == df['country_long'].apply(extracting_name)
print(count)
```

```python
#which country in african region has highest population
ans = df[df['continent'] == 'Africa']
ans[ans['population'] == ans['population'].max()]['country']
```
